### Using Your Own CSV Data

To use your own CSV file for customer segmentation:
1.  **Upload your CSV:** Click the folder icon on the left sidebar in Colab to open the file browser. Drag and drop your CSV file directly into the 'files' section.
2.  **Update the filename:** In the next code cell, replace `'your_data.csv'` with the actual name of your uploaded CSV file.
3.  **Check column names:** Ensure your CSV file contains a column named `'Amount'` (this is used for segmentation). If your amount column has a different name, please modify `'Amount'` in the code accordingly.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample data (you can replace with CSV upload)
data = {
    'Name': ['Rahul', 'Priya', 'Ankit', 'Sneha', 'Kiran'],
    'Amount': [1200, 700, 300, 1500, 450]
}

df = pd.DataFrame(data)

# Segmentation logic
def segment(amount):
    if amount > 1000:
        return 'High Value'
    elif amount > 500:
        return 'Medium Value'
    else:
        return 'Low Value'

df['Segment'] = df['Amount'].apply(segment)

print(df)

# Count segments
segment_counts = df['Segment'].value_counts()

# Plot
plt.figure()
segment_counts.plot(kind='bar')
plt.title("Customer Segmentation")
plt.xlabel("Segment")
plt.ylabel("Number of Customers")
plt.show()
```
